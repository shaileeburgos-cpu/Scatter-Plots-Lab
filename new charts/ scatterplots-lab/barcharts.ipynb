```python
# ============================================================
# 386.6.2 – Creating Bar Charts
# Completed by: Manuel
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Example 1: Simple Vertical Bar Chart
# ============================================================

CoursesName = ['C Course', 'C++ Course', 'Java Course', 'Python Course']
Student = [90, 60, 130, 190]

plt.figure(figsize=(8,5))
plt.bar(CoursesName, Student, color='green', width=0.3)
plt.xlabel('Courses Name')
plt.ylabel('Student Registrations Number')
plt.title('Number of Students in Different Courses')
plt.show()


# ============================================================
# Example 2: Simple Horizontal Bar Chart
# ============================================================

plt.figure(figsize=(8,5))
plt.barh(CoursesName, Student, color='green', height=0.3)
plt.xlabel('Student Registrations Number')
plt.ylabel('Courses Name')
plt.title('Number of Students in Different Courses')
plt.show()


# ============================================================
# Example 3.1: Stacked Bar Chart (Two Groups)
# ============================================================

StudentGroup1 = [90, 60, 130, 600]
StudentGroup2 = [100, 30, 160, 690]

plt.figure(figsize=(8,5))
plt.bar(CoursesName, StudentGroup1, color='green', width=0.3)
plt.bar(CoursesName, StudentGroup2, bottom=StudentGroup1, width=0.3)
plt.xlabel('Courses Name')
plt.ylabel('Student Registrations Number')
plt.title('Stacked Bar Chart – Two Groups')
plt.show()


# ============================================================
# Example 3.2: Stacked Bar Chart (Rice & Wheat)
# ============================================================

years = ['2016', '2017', '2018', '2019']
rice_price = [57, 50, 47, 30]
wheat_price = [43, 50, 53, 70]

plt.figure(figsize=(10,6))
plt.bar(years, rice_price, label='Rice')
plt.bar(years, wheat_price, bottom=rice_price, label='Wheat')
plt.xlabel('Years')
plt.ylabel('Price')
plt.title('The Price of Rice and Wheat')
plt.legend()
plt.show()


# ============================================================
# Example 3.3: Stacked Bar Chart (Three Groups)
# ============================================================

rice = np.array([57, 50, 47, 30])
wheat = np.array([43, 50, 53, 70])
mustard = np.array([10, 16, 13, 15])

plt.figure(figsize=(10,6))
plt.bar(years, rice, label='Rice')
plt.bar(years, wheat, bottom=rice, label='Wheat')
plt.bar(years, mustard, bottom=rice+wheat, label='Mustard Seeds')
plt.xlabel('Years')
plt.ylabel('Preference (%)')
plt.title('Rice, Wheat & Mustard Seeds Export Results')
plt.legend()
plt.show()


# ============================================================
# Example 4: Grouped Bar Chart
# ============================================================

width = 0.3
x = np.arange(len(CoursesName))

plt.figure(figsize=(10,6))
plt.bar(x - width/2, StudentGroup1, width, color='red', label='Student Group 1')
plt.bar(x + width/2, StudentGroup2, width, color='green', label='Student Group 2')
plt.xlabel('Courses Name')
plt.ylabel('Number of Students')
plt.title('Number of Students in Different Courses')
plt.xticks(x, CoursesName, rotation=45, ha='right')
plt.legend()
plt.show()


# ============================================================
# Example 5: Bar Chart Using Pandas
# ============================================================

data = {
    'CoursesName': ['C Programming Course', 'C++ Programming Course', 'Java Programming Course', 'Python Programming Course'],
    'Student': [90, 60, 130, 600]
}

df = pd.DataFrame(data)

plt.figure(figsize=(10,6))
plt.bar(df['CoursesName'], df['Student'], color='green')
plt.xlabel('Courses Name')
plt.ylabel('Number of Students')
plt.title('Number of Students in Different Courses')
plt.xticks(rotation=45, ha='right')
plt.show()


# ============================================================
# Example 6: Group Bar Chart Using Pandas
# ============================================================

data2 = {
    'CoursesName': ['C Programming Course', 'C++ Programming Course', 'Java Programming Course', 'Python Programming Course'],
    'Student Group 1': [90, 60, 130, 600],
    'Student Group 2': [100, 30, 160, 690]
}

df2 = pd.DataFrame(data2)

bar_width = 0.35
index = df2.index

plt.figure(figsize=(10,6))
plt.bar(index - bar_width/2, df2['Student Group 1'], bar_width, label='Student Group 1', color='blue')
plt.bar(index + bar_width/2, df2['Student Group 2'], bar_width, label='Student Group 2', color='orange')
plt.xlabel('Courses Name')
plt.ylabel('Number of Students')
plt.title('Number of Students in Different Courses')
plt.xticks(index, df2['CoursesName'], rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```
