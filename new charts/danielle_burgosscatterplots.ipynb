```python
import matplotlib.pyplot as plt
import numpy as np

# Example 1: Simple Scatter Plot
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.grid()
plt.scatter(x, y, color="red", s=90)
plt.title("Simple Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
# Example 2: Change the size of the dots
sizes = np.array([20,50,100,200,500,1000,60,90,10,300,600,800,75])

plt.scatter(x, y, color="red", s=sizes)
plt.title("Scatter Plot with Variable Marker Sizes")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
# Example 3: Adjust transparency
plt.scatter(x, y, s=sizes, color="red", alpha=0.5)
plt.title("Scatter Plot with Transparency")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
# Example 4: Combine color, size, and alpha
x = np.random.randint(100, size=100)
y = np.random.randint(100, size=100)
colors = np.random.randint(100, size=100)
sizes = 10 * np.random.randint(100, size=100)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='nipy_spectral')
plt.colorbar()
plt.title("Scatter Plot with Colormap, Size, and Transparency")
plt.xlabel("Random X")
plt.ylabel("Random Y")
plt.show()
```

```python
# Example 5: Compare two scatter plots

AgeGroupOne = range(40, 70)
Purchase1 = np.abs(np.random.randn(30) * 20)

plt.scatter(AgeGroupOne, Purchase1, s=200, alpha=0.6, label='40–69 years old')
plt.scatter(AgeGroupOne, Purchase1, color='#88c999')

AgeGroupTwo = range(20, 40)
Purchase2 = np.abs(np.random.randn(20) * 40)

plt.scatter(AgeGroupTwo, Purchase2, s=200, alpha=0.6, label='20–39 years old')
plt.scatter(AgeGroupTwo, Purchase2, color='#88c984')

plt.legend()
plt.xlabel("Age")
plt.ylabel("Average Purchase per Weekend")
plt.title("Comparison of Two Age Groups")
plt.show()
```

```python
# Example 6: Randomly generated scatter plot

rng = np.random.RandomState(0)
x = rng.randn(100)
y = rng.randn(100)

colors = rng.rand(100)
sizes = 1000 * rng.rand(100)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.3)
plt.title("Random Scatter Plot")
plt.xlabel("Random X")
plt.ylabel("Random Y")
plt.colorbar()
plt.show()
```

```python
# Example 7: Real-time housing data visualization
import pandas as pd
import matplotlib.pyplot as plt

housing_df = pd.read_csv("https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv")

plt.figure(figsize=(12, 8))
plt.scatter(
    x=housing_df['longitude'],
    y=housing_df['latitude'],
    alpha=0.4,
    s=housing_df['population'] / 100,
    c=housing_df['median_house_value'],
    cmap='jet'
)

plt.colorbar(label='Median House Value')
plt.legend(['Population'])
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('California Housing Data — Population & Median House Value')
plt.show()
```
